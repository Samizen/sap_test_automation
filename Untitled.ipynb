```python
import os
from openpyxl import load_workbook
import pandas as pd
```

```python
file_path = os.path.join("excel_name_change/excel_files/", "GPO-001 - Goods Purchase - CIF Catalog - Low Dollar.xlsx")
```

```python
wb = load_workbook(file_path)
```

```python
vars_to_change = {
    "Scenario":"__changed",
    "Environment":"env",
    "Enter":"_enter_" 
}
def read_through_df(df:pd.DataFrame):
    for var in vars_to_change:
        print(f"Changing {var} to {vars_to_change[var]}")
        df.replace('Scenario Number', "test")
    return df
```

```python
for sheet in wb.sheetnames:
    sheet_name=sheet
    ws = wb[sheet]
    df = pd.DataFrame(ws.values)
    for var in vars_to_change:
        print(f"Changing '{var}' to '{vars_to_change[var]}'")
        df.replace(var, vars_to_change[var], inplace=True,regex=True)
    for row_index, row in enumerate(df.values, start=1):
        for col_index, value in enumerate(row, start=1):
            ws.cell(row=row_index, column=col_index, value=value)
            
```
